```python
# Install required libraries if you haven't already:
# %pip install pandas plotly

import pandas as pd
import plotly.express as px

# 1. Load the generated sample CSV file
df = pd.read_csv("sample_sales_data.csv")

# 2. Generate an interactive line chart
fig = px.line(
    df, 
    x="Date", 
    y=["Sales", "Profit"], 
    title="Business Performance Over Time (2026)",
    labels={"value": "Amount ($)", "variable": "Metric"}
)

# 3. Render the interactive graph right inside VS Code
fig.show()
```
